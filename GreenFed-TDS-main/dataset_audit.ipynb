```python
# Task E4 — Duplicate check within chunks
# Note: This method does not detect duplicates across chunk boundaries.

import pandas as pd

def count_duplicates_in_chunks(path, chunksize=100_000):
    duplicate_count = 0

    for chunk in pd.read_csv(path, chunksize=chunksize, low_memory=False):
        duplicate_count += int(chunk.duplicated().sum())

    return duplicate_count

path1 = DATA_DIR / "Merged01.csv"
path2 = DATA_DIR / "Merged02.csv"

duplicates_01 = count_duplicates_in_chunks(path1)
duplicates_02 = count_duplicates_in_chunks(path2)

print("===== DUPLICATE CHECK =====")
print("Merged01.csv duplicates within chunks:", duplicates_01)
print("Merged02.csv duplicates within chunks:", duplicates_02)
print("Total counted within chunks:", duplicates_01 + duplicates_02)

print("\nLimitation: duplicates across chunk boundaries and")
print("duplicates shared between files are not included.")
```

```python
# Task E5 — Prepare full-file schema and quality report

import numpy as np
import pandas as pd

def audit_schema_quality(path, filename):
    missing_counts = {}
    infinite_counts = {}
    dtypes = {}
    columns = None

    for chunk in pd.read_csv(
        path,
        chunksize=100_000,
        low_memory=False
    ):
        if columns is None:
            columns = list(chunk.columns)
            dtypes = {col: str(chunk[col].dtype) for col in columns}
            missing_counts = {col: 0 for col in columns}
            infinite_counts = {col: 0 for col in columns}

        for col in columns:
            missing_counts[col] += int(chunk[col].isna().sum())

        numeric = chunk.select_dtypes(include=[np.number])

        for col in numeric.columns:
            infinite_counts[col] += int(np.isinf(numeric[col]).sum())

    return [
        {
            "file": filename,
            "position": i,
            "column": col,
            "dtype_first_chunk": dtypes[col],
            "missing_count": missing_counts[col],
            "infinite_count": infinite_counts[col],
        }
        for i, col in enumerate(columns, start=1)
    ]

quality_rows = []

quality_rows.extend(
    audit_schema_quality(DATA_DIR / "Merged01.csv", "Merged01.csv")
)
quality_rows.extend(
    audit_schema_quality(DATA_DIR / "Merged02.csv", "Merged02.csv")
)

schema_quality = pd.DataFrame(quality_rows)

schema_quality.to_csv("schema_and_quality.csv", index=False)

print("Created schema_and_quality.csv")
print("Rows in report:", len(schema_quality))
print(schema_quality.head())
```

```python
print("Missing values:")
print(schema_quality.groupby("file")["missing_count"].sum())

print("\nInfinite values:")
print(schema_quality.groupby("file")["infinite_count"].sum())

print("\nTotal report rows:", len(schema_quality))
```

```python
# Task F — Compare file compatibility

schema_01 = schema_quality[
    schema_quality["file"] == "Merged01.csv"
].sort_values("position")

schema_02 = schema_quality[
    schema_quality["file"] == "Merged02.csv"
].sort_values("position")

same_columns = (
    schema_01["column"].tolist()
    == schema_02["column"].tolist()
)

same_count = len(schema_01) == len(schema_02)

label_in_both = (
    "Label" in schema_01["column"].tolist()
    and "Label" in schema_02["column"].tolist()
)

same_dtypes = (
    schema_01["dtype_first_chunk"].tolist()
    == schema_02["dtype_first_chunk"].tolist()
)

print("===== TASK F: COMPATIBILITY CHECK =====")
print("Same column names and order:", same_columns)
print("Same column count:", same_count)
print("Label exists in both files:", label_in_both)
print("Same data types by column:", same_dtypes)

print("\n===== DATA TYPE DIFFERENCES =====")

comparison = pd.DataFrame({
    "column": schema_01["column"].tolist(),
    "Merged01_dtype": schema_01["dtype_first_chunk"].tolist(),
    "Merged02_dtype": schema_02["dtype_first_chunk"].tolist()
})

differences = comparison[
    comparison["Merged01_dtype"] != comparison["Merged02_dtype"]
]

if differences.empty:
    print("No data type differences found.")
else:
    print(differences.to_string(index=False))
```
